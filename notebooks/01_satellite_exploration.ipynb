# 01 — Satellite SAR Imagery Exploration
Exploration of Sentinel-1 C-SAR Ground Range Detected (GRD) VV measurement rasters, image statistics, and spatial coverage for Case 001.


```python
from src.common.case_loader import load_case
from src.common.paths import resolve_path
import rasterio
import numpy as np
import matplotlib.pyplot as plt
```

```python
case = load_case("case_001")
sat_file = case.satellite_files["measurement_raster_vv"]
print(f"Loading SAR raster: {sat_file}")
```

```python
with rasterio.open(sat_file) as src:
    print(f'Dimensions: {src.shape}')
    print(f'CRS: {src.crs}')
    print(f'Bounds: {src.bounds}')
    data = src.read(1)
```

```python
valid = data[data > 0]
print(f'Valid pixels: {len(valid)} / {data.size} ({len(valid)/data.size*100:.1f}%)')
print(f'DN range: [{valid.min()} .. {valid.max()}], Median DN: {np.median(valid):.1f}')
```
